```python
import pandas as pd

# Input and Output file paths
input_file = r"C:\Users\USER\Downloads\12. IVDL KITWE -2033 - Dec 2025.xlsx"
output_file = r"12. IVDL KITWE -2033 - Dec 2025.xlsx"

# Read all sheets into a dictionary
all_sheets = pd.read_excel(input_file, sheet_name=None)

# Merge all sheets
merged_df = pd.concat(all_sheets.values(), ignore_index=True)

# Save to new Excel file
merged_df.to_excel(output_file, index=False)

print("All sheets merged successfully!")
```
